# Write to bronze layer

```python
def write_to_bronze(data_path, table_name):

    df = spark.read.option("mergeSchema", "true").csv(data_path, header=True, inferSchema=True)

    df.write.mode("overwrite").saveAsTable("workspace.bronze." + table_name)

# Write crm files to bronze
write_to_bronze("/Volumes/workspace/bronze/source_systems/source_crm/cust_info.csv", "crm_cust_info")
write_to_bronze("/Volumes/workspace/bronze/source_systems/source_crm/prd_info.csv", "crm_prd_info")
write_to_bronze("/Volumes/workspace/bronze/source_systems/source_crm/sales_details.csv", "crm_sales_details")

# Write erp files to bronze
write_to_bronze("/Volumes/workspace/bronze/source_systems/source_erp/CUST_AZ12.csv", "erp_cust_az12")
write_to_bronze("/Volumes/workspace/bronze/source_systems/source_erp/LOC_A101.csv", "erp_loc_a101")
write_to_bronze("/Volumes/workspace/bronze/source_systems/source_erp/PX_CAT_G1V2.csv", "erp_px_cat_g1v2")
```
